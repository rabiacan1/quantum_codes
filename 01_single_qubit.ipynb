```python
from qiskit.quantum_info import Statevector
from qiskit.visualization import plot_bloch_multivector
import numpy as np
from numpy import sqrt


#definition of the vectors |0> and |1>
ket0= Statevector([1,0])
ket1=Statevector([0,1])
print("ket0 and ket1 vectors are defined succesfully...")

#superposition of the vectors
psi=ket0/sqrt(2)+ ket1/sqrt(2)
display(psi.draw("latex"))

#Measuring the wave function psi
outcome,collapsed_state=psi.measure()
print(f"Measured value:{outcome}")
print("Collapsed state after measuring:")
display(collapsed_state.draw("latex"))


#drawing the image of 3D Bloch sphere according to our definitions
plot_bloch_multivector(psi)
```
